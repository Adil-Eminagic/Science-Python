```python
print(len(image))
print(len(image[0]))
print(len(image[0][0]))
```

```python
darker_image = np.array(image, copy=True)

n = 200

for i in range(len(darker_image)):
    for j in range(len(darker_image[0])):
        for k in range(len(darker_image[0][0])):
            value = darker_image[i][j][k]
            if value >= n:
                darker_image[i][j][k] = value - n
            else:
                darker_image[i][j][k] = 0

            
plt.imshow(darker_image)

#plt.imshow(image)

plt.show()
```

```python
brighter_image = np.array(image, copy=True)

n = 10

for i in range(len(brighter_image)):
    for j in range(len(brighter_image[0])):
        for k in range(len(brighter_image[0][0])):
            value = brighter_image[i][j][k]
            if n + value <= 255:
                brighter_image[i][j][k] = value + n
            else:
                brighter_image[i][j][k] = 255

            
plt.imshow(brighter_image)


plt.show()
```
