```python
import sympy as sp
```

## Sum with sigma Σ sign

```python
w, b, num_items = sp.symbols('w b n')
i = sp.symbols('i', integer=True)
x = sp.IndexedBase('x')
y = sp.IndexedBase('y')

L = sp.Sum((w*x[i] + b - y[i])**2, (i, 1, num_items)) / num_items

display(sp.Eq(sp.Symbol('L'), L))

L_func = sp.Function('L')
display(sp.Eq(L_func(w, b), L))
```

## Partial deivatives and Gradient

```python
sp.init_printing()

w, b = sp.symbols('w b')

L = ((b - 1)**2 + (w + b - 3)**2) / 2

display(sp.Eq(sp.Function('L')(w, b), L))
```

```python
dL_dw = sp.diff(L, w)
dL_db = sp.diff(L, b)

display(sp.Eq(sp.Derivative(L, w), dL_dw))
display(sp.Eq(sp.Derivative(L, b), dL_db))
```

```python
gradient = sp.Matrix([dL_dw, dL_db])

display(sp.Eq(
    sp.Symbol(r'\nabla L'),
    gradient,
    evaluate=False
))
```
