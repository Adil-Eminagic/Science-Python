```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
from IPython.display import display, Math
```

# Poisson distribution

It is discrete probability distribution that expresses the likelihood of a given number of events occurring within a fixed interval of time, space, distance, or volume.

- **k** = the exact number of occurrences you want to find the probability for (\(k = 0, 1, 2, \dots\))
- **λ (lambda)** = the average or expected number of events in the interval

Exampel: average number of calls per one hour is 3. With Poisson distribtuion you find probability of getting any number of calls in one hour

```python
lambd, k = sp.symbols('λ k')

P = sp.exp(- lambd) * lambd**k / sp.factorial(k)

display(P)
display(Math(r"P(X=k) = " + sp.latex(P)))
display(sp.Eq(sp.Function('P')(k), P))
```

```python
result = P.subs({lambd: 3, k: 2})

display(result)
display(result.evalf())
```

```python
nCalls = np.arange(0,11)
avg_calls_hour= 3

probabilites = [ P.subs({lambd: avg_calls_hour, k: k_value}) for k_value in nCalls] 

plt.scatter(nCalls, probabilites)
# plt.plot(nCalls, probabilites)

plt.title(f'Possion distribution with avg calls/hour of {avg_calls_hour}', fontsize= 17)
plt.xlabel('Number of calls')
plt.ylabel('Probabilities for each num of calls/hour')
plt.show()
```

# Exponetial distibution

Exponetial distribution is like continous version of Poisson distribution. An Exponential distribution is a continuous probability distribution that models the time, distance, or space between independent events happening at a constant average rate.

- **Rate Parameter (λ)**: The average number of events per unit of time or space (λ > 0)


```python
lambd, x = sp.symbols('λ x')

f = lambd * sp.exp(-x * lambd)

display(P)
display(sp.Eq(sp.Function('f')(x), f))
```
