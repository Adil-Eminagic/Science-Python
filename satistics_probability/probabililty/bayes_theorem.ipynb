```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2
```

```python
A = { 1 , 2 , 3 , 4 , 5 , 6}
B = { 4 , 5 , 6 , 7 , 8 , 9}

v = venn2([A, B], set_labels = ('A', 'B'))

plt.title("Venn Diagram of Two Sets A and B")
plt.show()
```

```python
v.get_label_by_id('10').set_text('A only')
v.get_label_by_id('01').set_text('B only')
v.get_label_by_id('11').set_text('A and B')

plt.show()
```
