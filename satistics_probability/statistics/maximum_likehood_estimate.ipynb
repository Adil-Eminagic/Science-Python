# Maximum Likehood Estimate 

### Using derivatives

Since estimation of parameter p becomes function, and we need to find max value, for that we can use derivative. First derivative show us local minimum and maximums, and second derivative shows as are those minimum or maximums

```python
heads = 55
toss_number = 100
cominations = math.comb(toss_number, heads)

print(cominations)

# I need to find p

def bionomail_estimate(p):
    return cominations * p**heads * (1 - p)**(toss_number - heads)
```

```python
probabilites = np.linspace(0,1, 21)

print(probabilites)
```

```python
estimates = bionomail_estimate(probabilites)
```

```python
plt.scatter(probabilites, estimates)
plt.plot(probabilites, estimates)

plt.title(f'Distribution of estimates for p paramter in binomial distribtution 55/100', fontsize= 17)
plt.xlabel('Value od p parameter')
plt.ylabel('Nuber of ')
plt.show()
```
