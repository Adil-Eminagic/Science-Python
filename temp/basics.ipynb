## RGB and HSI image reperesenation

Screen consists of many tiny squares called **pixels**. It's full name is **picture element** and it is the smallest single unit or building block of a digital image or display. Number of pixels determine resolution of screeen. More pixels better result in better resoultion. Each pixel has it's unique color which is represented with amounts of red, gereen and blue color whcih is **RGB** represenation. Values goes from 0-255, which is 2^8, so we need 8 bits to represent each of three main colors.
- rgb(0,0,0) is balck color
- rgb(255,255,255) is white color
- 255 of one color and 0 for the remaining two result in pure main colors (red, green or blue)
- if all three values are same we all shades of gray

Another repreesentation is HSI (Hue Saturation Intensity), which is expalained in following videos:
- https://www.youtube.com/watch?v=Ceur-ARJ4Wc
- https://www.youtube.com/watch?v=pjo3wP_yt2A

- **Binary images** have pixels values only 0 and 1
- **Grayscale image** has only only one gray color from values 0-255
- **Colored image** has pixels has three values for red, green and blue colors

### Displaying images

```python
# Create a small grayscale image
image = np.array([
    [20, 50, 100],
    [80, 130, 180],
    [150, 210, 255]
], dtype=np.uint8)

# Display it
plt.imshow(image, cmap="gray", vmin=0, vmax=255)
plt.colorbar(label="Pixel intensity")
plt.show()
```
